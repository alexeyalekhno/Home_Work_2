```python
class Chocolate:
  def __init__(self, price: float, additive: str = None):
    if price <=0:
      raise ValueError ('Цена батончика должна быть больше 0')
    self.price = price
    self.additive = additive
  def show_my_chocolate(self):
    if self.additive:
      print(f'Шоколадный батончик, добавка: {self.additive}')
    else:
      print('Обычный шоколадный батончик')

  def __str__(self):
    return f'Шоколадный батончик, {self.price} рублей'

```

```python
chocolate = Chocolate(500, 'миндаль')
```

```python
try:
  chocolate.show_my_chocolate()
except ValueError as e:
  print(e)
```
